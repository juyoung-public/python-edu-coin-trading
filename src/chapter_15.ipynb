## 3. 부동소수점 오차와 안전한 비교

컴퓨터는 `0.1`을 정확하게 표현할 수 없기 때문에 오차가 발생할 수 있습니다.

```python
import math

value = 0.1 + 0.1 + 0.1
print(value)
print(value == 0.3)
print(math.isclose(value, 0.3))
print(math.isclose(value, 0.3, rel_tol=1e-12, abs_tol=0.0))
print(round(math.pi, 2))
```

## 4. `Decimal`과 `Fraction`

금융 계산이나 정확한 분수 표현은 `Decimal`과 `Fraction`이 적합합니다.

```python
from decimal import Decimal
from fractions import Fraction

print(Decimal("0.1") + Decimal("0.2"))
print(Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))

third = Fraction(1, 3)
print(third + third + third)
print(Fraction.from_float(0.1))
```

## 5. `math.fsum()`과 합계의 정밀도

많은 실수를 더할 때는 `math.fsum()`이 더 안정적인 결과를 내는 경우가 많습니다.

```python
import math

values = [0.1] * 10
print(sum(values))
print(math.fsum(values))

returns = [0.0001, -0.0001, 0.0002, -0.0002]
print(sum(returns))
print(math.fsum(returns))
```

## 8. 실습 검증

이 장의 핵심 개념을 실제로 확인합니다.

```python
import math
from decimal import Decimal
from fractions import Fraction

assert math.isclose(0.1 + 0.1 + 0.1, 0.3)
assert Decimal("0.1") + Decimal("0.2") == Decimal("0.3")
assert Fraction(1, 3) + Fraction(1, 3) + Fraction(1, 3) == Fraction(1, 1)
assert math.fsum([0.1] * 10) > 0.0
print("Chapter 15 실습 검증 통과")
```
