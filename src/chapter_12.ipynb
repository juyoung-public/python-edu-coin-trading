## 3. 명령줄 인자: `sys.argv`와 `argparse`

스크립트에 전달된 인자는 문자열로 처리되고, 선택 옵션은 `argparse`로 정리할 수 있습니다.

```python
import sys
import argparse

print("실행 파일:", sys.argv[0])
print("전달된 인자:", sys.argv[1:])

symbol = sys.argv[1] if len(sys.argv) > 1 else "BTC"
price_text = sys.argv[2] if len(sys.argv) > 2 else "0"
price = float(price_text)
print(f"{symbol}: {price:,.0f}원")

parser = argparse.ArgumentParser(description="코인 가격을 표시합니다.")
parser.add_argument("symbol", help="코인 심볼")
parser.add_argument("--price", type=float, default=0, help="현재 가격")
args = parser.parse_args([] if len(sys.argv) > 1 else ["BTC"])
print(f"argparse 예제: {args.symbol}: {args.price:,.0f}원")
```

## 5. 문자열 패턴 검색: `re`

정규 표현식은 문자열에서 구조화된 패턴을 찾을 때 강력합니다.

```python
import re

text = "BTC:105000000 ETH:3500000 XRP:800"
prices = re.findall(r"[A-Z]+:\d+", text)
print(prices)

record = "symbol=BTC price=105000000"
match = re.search(r"symbol=(\w+) price=(\d+)", record)
if match:
    symbol, price = match.groups()
    print(symbol, int(price))
```

## 6. 수학과 통계

`math`, `random`, `statistics`는 기본적인 계산과 샘플링을 도와줍니다.

```python
import math
import random
import statistics

print(math.pi)
print(math.sqrt(16))
print(math.floor(105.67), math.ceil(105.67))
print(math.isclose(0.1 + 0.2, 0.3))

random.seed(42)
print(random.choice(["BTC", "ETH", "XRP"]))
print(random.sample(range(100), 5))

returns = [1.2, -0.5, 2.1, 0.8, -1.0]
print(statistics.mean(returns))
print(statistics.median(returns))
print(statistics.pvariance(returns))
```

## 7. 인터넷 접근: `urllib.request`

네트워크 URL에서 데이터를 가져오고 응답 결과를 확인할 수 있습니다.

```python
from urllib.request import urlopen

url = "https://docs.python.org/3/"
with urlopen(url, timeout=10) as response:
    print("상태 코드:", response.status)
    print("콘텐츠 유형:", response.headers.get("Content-Type"))

with urlopen(url, timeout=10) as response:
    for line in response:
        text = line.decode("utf-8", errors="replace")
        if "Python" in text:
            print(text.strip())
            break
```

## 8. 날짜와 시간: `datetime`

날짜와 시간을 생성하고 포맷을 변환하는 기본 도구를 알아봅니다.

```python
from datetime import date, datetime, timedelta

print("오늘:", date.today())
now = datetime.now()
print("현재 시간:", now)
print("일주일 뒤:", now + timedelta(days=7))

formatted = now.strftime("%Y-%m-%d %H:%M:%S")
print("포맷 문자열:", formatted)
parsed = datetime.strptime("2026-07-31 09:30:00", "%Y-%m-%d %H:%M:%S")
print("파싱 결과:", parsed)

start = date(2026, 7, 1)
end = date(2026, 7, 31)
print("날짜 차이:", (end - start).days)
```

## 9. 데이터 압축: `zlib`와 `zipfile`

압축은 대용량 텍스트와 파일 묶음을 효율적으로 다루는 데 유용합니다.

```python
import zlib
from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED

raw_data = b"BTC,105000000\nETH,3500000\n" * 10
compressed = zlib.compress(raw_data)
restored = zlib.decompress(compressed)
print("원본 길이:", len(raw_data))
print("압축 길이:", len(compressed))
print("복원 성공:", restored == raw_data)
print("CRC32:", zlib.crc32(raw_data))

text = "코인 가격 데이터입니다." * 10
compressed_text = zlib.compress(text.encode("utf-8"))
restored_text = zlib.decompress(compressed_text).decode("utf-8")
print("텍스트 복원 성공:", restored_text == text)

sample_path = Path("sample_prices.txt")
sample_path.write_text("BTC,105000000\nETH,3500000\n", encoding="utf-8")
with ZipFile("sample_prices.zip", "w", compression=ZIP_DEFLATED) as archive:
    archive.write(sample_path, arcname=sample_path.name)
with ZipFile("sample_prices.zip") as archive:
    print("ZIP 파일 목록:", archive.namelist())
```

## 10. 성능 측정: `timeit`

`timeit`은 반복문과 컴프리헨션 같은 구현을 비교하는 데 적합합니다.

```python
import timeit

loop_time = timeit.timeit(
    "result = []\nfor value in range(100):\n    result.append(value * value)",
    number=1_000,
)
comprehension_time = timeit.timeit(
    "result = [value * value for value in range(100)]",
    number=1_000,
)
print(f"반복문: {loop_time:.5f}초")
print(f"컴프리헨션: {comprehension_time:.5f}초")
```

## 11. 품질 관리: `doctest`와 `unittest`

문서 예제와 테스트를 함께 관리하면 코드와 설명이 맞는지 쉽게 확인할 수 있습니다.

```python
import doctest
import unittest


def average(values):
    """숫자 목록의 평균을 반환합니다.

    >>> average([20, 30, 70])
    40.0
    >>> average([1, 5, 7])
    4.333333333333333
    """
    return sum(values) / len(values)


print(doctest.testmod())


def calculate_fee(amount, fee_rate=0.0004):
    return amount * fee_rate


class TestTradingFunctions(unittest.TestCase):
    def test_fee(self):
        self.assertEqual(calculate_fee(10_000), 4.0)

    def test_invalid_amount(self):
        with self.assertRaises(ValueError):
            if -1 < 0:
                raise ValueError("금액은 음수일 수 없습니다.")


suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestTradingFunctions)
result = unittest.TextTestRunner(verbosity=0).run(suite)
print(f"성공: {result.wasSuccessful()}")
```
