## 1. 리스트와 주요 메서드

리스트는 순서가 있고 수정할 수 있는 자료구조입니다. 가격 목록을 추가, 수정, 삭제해 보겠습니다.

```python
prices = [100, 105, 103]

prices.append(110)
prices.extend([98, 115])
prices.insert(1, 102)
prices[0] = 101

print("현재 가격:", prices)
print("마지막 가격:", prices.pop())
print("가격 개수:", len(prices))
print("정렬된 복사본:", sorted(prices))
print("원본 리스트:", prices)

orders = ["BTC", "ETH"]
orders.append("XRP")
orders.remove("ETH")
print("주문 목록:", orders)
```

## 2. 스택과 큐

스택은 마지막에 들어온 항목을 먼저 꺼내고, 큐는 먼저 들어온 항목을 먼저 꺼냅니다. 큐에는 `deque`를 사용하는 것이 효율적입니다.

```python
stack = ["매수 신호", "가격 확인"]
stack.append("주문 제출")
print("스택에서 꺼냄:", stack.pop())
print("남은 스택:", stack)

from collections import deque

queue = deque(["BTC 주문", "ETH 주문"])
queue.append("XRP 주문")
print("큐에서 꺼냄:", queue.popleft())
print("남은 큐:", queue)
```

## 3. 리스트 컴프리헨션과 `del`

리스트 컴프리헨션은 반복문으로 새 리스트를 만드는 표현을 간결하게 작성합니다. `del`은 인덱스나 슬라이스로 항목을 삭제합니다.

```python
prices = [100, 105, 110, 98]
doubled = [price * 2 for price in prices]
rise_prices = [price for price in prices if price >= 100]

print("두 배 가격:", doubled)
print("100 이상 가격:", rise_prices)

coins = ["BTC", "ETH", "XRP"]
timeframes = ["1d", "1h"]
requests = [
    (coin, timeframe)
    for coin in coins
    for timeframe in timeframes
    if coin != "XRP"
]
print("요청 목록:", requests)

working_prices = [100, 105, 110, 98, 102]
del working_prices[0]
del working_prices[1:3]
print("삭제 후 가격:", working_prices)
```

## 4. 튜플과 시퀀스 언패킹

튜플은 생성 후 항목을 바꿀 수 없는 불변 시퀀스입니다. 여러 값을 하나의 기록으로 묶거나 함수에서 여러 값을 반환할 때 사용할 수 있습니다.

```python
market_data = ("BTC", 105_000_000, 2.5)
symbol, price, change_rate = market_data
print(symbol, price, change_rate)

empty = ()
single = ("BTC",)
print("튜플 길이:", len(empty), len(single))

first, *middle, last = [1, 2, 3, 4, 5]
print("첫 값:", first)
print("중간 값:", middle)
print("마지막 값:", last)

matrix = [[1, 2, 3], [4, 5, 6]]
transposed = list(zip(*matrix))
print("전치 행렬:", transposed)
```

## 5. 집합과 집합 연산

집합은 중복을 제거하고 멤버십을 빠르게 확인할 때 사용합니다. 합집합, 교집합, 차집합을 거래 시장 비교에 적용해 보겠습니다.

```python
symbols = {"BTC", "ETH", "BTC", "XRP"}
print("중복 제거 심볼:", symbols)
print("ETH 지원:", "ETH" in symbols)

btc_markets = {"KRW", "USDT", "USD"}
eth_markets = {"KRW", "USDT", "EUR"}

print("합집합:", btc_markets | eth_markets)
print("교집합:", btc_markets & eth_markets)
print("BTC에만 있는 시장:", btc_markets - eth_markets)
print("한쪽에만 있는 시장:", btc_markets ^ eth_markets)

letters = {letter for letter in "abracadabra" if letter not in "abc"}
print("집합 컴프리헨션:", letters)
```

## 6. 딕셔너리와 컴프리헨션

딕셔너리는 키와 값의 쌍으로 데이터를 저장합니다. 코인 심볼을 키로 사용해 가격과 변동률을 조회해 보겠습니다.

```python
market = {
    "symbol": "BTC",
    "price": 105_000_000,
    "change_rate": 2.5,
}

print("거래량:", market.get("volume", 0))

market["volume"] = 1_500
market["price"] = 106_000_000

print("심볼:", market["symbol"])
print("거래량:", market.get("volume", 0))
print("없는 키:", market.get("missing"))

for key, value in market.items():
    print(f"{key}: {value}")

all_prices = {"BTC": 105, "ETH": 3, "XRP": 0.5}
expensive = {
    symbol: price
    for symbol, price in all_prices.items()
    if price >= 10
}
print("10 이상 가격:", expensive)

symbols = ["BTC", "ETH", "XRP"]
prices = [105, 3, 0.5]
price_table = dict(zip(symbols, prices))
print("가격표:", price_table)
```

## 7. 자료구조 순회 기법

`items()`는 딕셔너리의 키와 값을 함께 제공하고, `enumerate()`는 순번과 값을, `zip()`은 여러 시퀀스의 항목을 같은 위치끼리 묶습니다.

```python
coins = ["BTC", "ETH", "XRP"]

for rank, coin in enumerate(coins, start=1):
    print(f"{rank}위: {coin}")

symbols = ["BTC", "ETH", "XRP"]
prices = [105, 3, 0.5]
changes = [2.5, -1.2, 0.3]

for symbol, price, change in zip(symbols, prices, changes):
    print(f"{symbol}: 가격={price}, 변동률={change}%")

print("역순:", list(reversed(coins)))
print("정렬:", sorted(coins))
print("중복 제거 후 정렬:", sorted(set(["BTC", "ETH", "BTC", "XRP"])))
```

## 8. 안전한 자료 정리와 조건식

반복 중인 리스트를 직접 삭제하면 항목을 건너뛸 수 있습니다. 새 리스트를 만들고 `and`, `or`, `is None`을 이용해 데이터를 안전하게 정리합니다.

```python
prices = [100, None, 105, None, 110]
valid_prices = [price for price in prices if price is not None]
print("유효한 가격:", valid_prices)

price = None
if price is not None and price > 100:
    print("가격 조건을 만족합니다.")
else:
    print("가격이 없거나 조건을 만족하지 않습니다.")

name = ""
display_name = name or "이름 없음"
print("표시 이름:", display_name)

change_rate = 2.5
print("완만한 상승:", 0 < change_rate <= 5)
```

## 9. 시퀀스 비교

리스트, 튜플, 문자열은 같은 종류끼리 앞에서부터 비교합니다. 처음으로 서로 다른 항목이 전체 비교 결과를 결정합니다.

```python
print((1, 2, 3) < (1, 2, 4))
print([1, 2, 3] == [1, 2, 3])
print("ABC" < "C")
print((1, 2) < (1, 2, -1))
print((1, 2, 3) == (1.0, 2.0, 3.0))
```
