## 1. 왜 가상 환경이 필요한가?

프로젝트마다 패키지 버전이 다를 수 있으므로 전역 환경 하나로는 관리하기 어렵습니다.

````python
import sys
from pathlib import Path

print("Python 실행 파일:", sys.executable)
print("sys.prefix:", sys.prefix)
print("sys.base_prefix:", sys.base_prefix)
print("가상 환경 여부:", sys.prefix != sys.base_prefix)
print("프로젝트 루트 존재:", Path(".").exists())
````

## 2. 가상 환경 만들기

Windows PowerShell 예시:

```powershell
python -m venv .venv
```

macOS/Linux 예시:

```bash
python3 -m venv .venv
```

````python
import subprocess
from pathlib import Path

venv_dir = Path(".venv")
print("가상 환경 디렉터리 존재 여부:", venv_dir.exists())
print("가상 환경 폴더 안의 Python 실행 파일:", (venv_dir / "Scripts" / "python.exe").exists() or (venv_dir / "bin" / "python").exists())
````

## 3. 가상 환경 활성화

PowerShell에서:

```powershell
.\.venv\Scripts\Activate.ps1
```

macOS/Linux에서:

```bash
source .venv/bin/activate
```

활성화된 상태에서는 `python`과 `pip`가 현재 프로젝트의 환경을 가리킵니다.

````python
import os

print("VIRTUAL_ENV가 설정되어 있나요?", "VIRTUAL_ENV" in os.environ)
print("현재 작업 디렉터리:", os.getcwd())
````

## 4. 패키지 설치와 관리

`python -m pip install`은 전역 환경 대신 현재 환경에 패키지를 설치하는 표준적인 방식입니다.

````python
import sys

for command in ["pip", "requests", "pandas", "matplotlib"]:
    print(command, "은(는) 실행 시점에 확인할 수 있습니다.")
print("설치용 명령 예시:", "python -m pip install pandas matplotlib")
print("리스트 확인 예시:", "python -m pip list")
print("고정 의존성 예시:", "python -m pip freeze > requirements.txt")
````

## 5. `requirements.txt`와 재현 가능한 환경

`requirements.txt`를 저장해 두면 다른 컴퓨터에서도 같은 환경을 만들 수 있습니다.

````python
from pathlib import Path

requirements = Path("requirements.txt")
print("requirements.txt 파일 존재:", requirements.exists())
print("권장 내용 예시:")
print("pandas>=2.0")
print("matplotlib>=3.8")
print("requests>=2.31")
````

## 6. 실습 검증

가상 환경 개념과 의존성 관리의 기본 원칙이 지켜지는지 확인합니다.

````python
import sys
from pathlib import Path

assert sys.executable
assert isinstance(sys.prefix, str)
assert isinstance(sys.base_prefix, str)
assert Path(".").exists()
print("Chapter 14 실습 검증 통과")
````
