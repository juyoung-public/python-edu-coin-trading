### 3.1 조건문과 논리 연산

`if`, `elif`, `else`는 조건에 따라 다른 코드를 실행합니다.

```python
price = 105_000_000
target_price = 100_000_000
change_rate = 2.5
volume = 1_500

if price > target_price and volume >= 1_000:
    signal = "매수 관심"
elif change_rate > 0:
    signal = "상승 관찰"
else:
    signal = "대기"

print(signal)

supported_markets = {"BTC", "ETH", "XRP"}
print("BTC" in supported_markets)
```

### 3.2 `for`, `range()`, `break`, `continue`

`for`는 반복 가능한 객체를 순회하고, `range()`는 숫자 범위를 만듭니다. `break`는 반복을 끝내며 `continue`는 현재 반복만 건너뜁니다.

```python
prices = [98, None, 101, 103, 99]
valid_prices = []

for day in range(len(prices)):
    price = prices[day]
    if price is None:
        continue
    valid_prices.append(price)
    print(f"{day}일차 가격: {price}")
    if price >= 103:
        print("목표 가격에 도달했습니다.")
        break

print("유효한 가격:", valid_prices)
```

### 3.3 반복문의 `else`와 `pass`

반복문 `else`는 `break` 없이 반복이 끝났을 때 실행됩니다. `pass`는 아직 구현하지 않은 블록의 자리 표시자입니다.

```python
coins = ["BTC", "ETH", "XRP"]

for coin in coins:
    if coin == "DOGE":
        print("거래 대상을 찾았습니다.")
        break
else:
    print("거래 대상이 없습니다.")


def future_indicator():
    pass

print("미구현 함수의 반환값:", future_indicator())
```

### 3.4 `match` 패턴 매칭

Python 3.10부터 `match`를 사용할 수 있습니다. 아래 셀을 실행해 상태 코드에 따른 메시지를 확인하세요.

```python
def explain_status(status):
    match status:
        case 200:
            return "정상 처리"
        case 400 | 401 | 403:
            return "요청 또는 인증 오류"
        case 404:
            return "대상을 찾을 수 없음"
        case _:
            return "알 수 없는 상태"


print(explain_status(200))
print(explain_status(500))


def describe_point(point):
    match point:
        case (0, 0):
            return "원점"
        case (x, y) if x == y:
            return "대각선 위의 점"
        case (x, y):
            return f"x={x}, y={y}"
        case _:
            return "점이 아닙니다"


print(describe_point((3, 3)))
print(describe_point((2, 5)))
```

### 3.5 함수 정의와 반환값

함수는 반복해서 사용할 작업을 이름으로 묶습니다. `return`은 계산 결과를 호출한 곳으로 돌려줍니다.

```python
def fibonacci_list(limit):
    """limit보다 작은 Fibonacci 수를 리스트로 반환합니다."""
    result = []
    first, second = 0, 1

    while first < limit:
        result.append(first)
        first, second = second, first + second

    return result


print(fibonacci_list(100))
print(fibonacci_list(0))  # 빈 리스트
```

### 3.6 기본값 인자, 키워드 인자, 특수 매개변수

기본값을 사용하면 인자를 생략할 수 있고, 키워드 인자는 호출 의도를 명확하게 합니다. `/` 앞은 위치 전용, `*` 뒤는 키워드 전용 매개변수입니다.

```python
def create_order(symbol, /, amount=10_000, *, order_type="market"):
    """거래 주문 정보를 딕셔너리로 만듭니다."""
    return {
        "symbol": symbol,
        "amount": amount,
        "order_type": order_type,
    }


print(create_order("BTC"))
print(create_order("ETH", amount=20_000, order_type="limit"))


def summarize_prices(*prices, **metadata):
    if not prices:
        return {"count": 0, "average": None, "metadata": metadata}
    return {
        "count": len(prices),
        "average": sum(prices) / len(prices),
        "metadata": metadata,
    }


print(summarize_prices(100, 105, 103, symbol="BTC", timeframe="1d"))
```

### 3.7 인자 목록 풀기와 람다

리스트 앞의 `*`와 딕셔너리 앞의 `**`는 함수 호출 시 인자를 풀어 전달합니다. 람다는 짧은 함수를 한 줄로 표현합니다.

```python
order_args = ["XRP", 5_000]
order_options = {"order_type": "limit"}
print(create_order(*order_args, **order_options))

pairs = [(3, "three"), (1, "one"), (2, "two")]
pairs.sort(key=lambda pair: pair[1])
print(pairs)

add_fee = lambda amount: amount * 1.0004
print(f"수수료 포함 금액: {add_fee(10_000):.2f}")
```

### 3.8 독스트링과 함수 어노테이션

독스트링은 함수의 목적을 설명하고, 어노테이션은 예상 타입 정보를 표시합니다. 어노테이션은 자동 타입 검사를 수행하지 않습니다.

```python
def calculate_return(buy_price: float, sell_price: float) -> float:
    """매수가와 매도가로 단순 수익률을 계산합니다."""
    return (sell_price - buy_price) / buy_price


print(calculate_return(100.0, 110.0))
print(calculate_return.__doc__)
print(calculate_return.__annotations__)
```
